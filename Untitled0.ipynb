```python
# Nama            : Kharis Ramadila
# NIM             : 422501033
# Tanggal         : 27 September 2026
# Tujuan Program  : Melakukan berbagai operasi aritmetika
#                   terhadap dua bilangan bulat.

print("=== PROGRAM OPERASI ARITMETIKA ===")

bilangan_1 = int(input("Masukkan bilangan bulat pertama: "))
bilangan_2 = int(input("Masukkan bilangan bulat kedua: "))

penjumlahan = bilangan_1 + bilangan_2
pengurangan = bilangan_1 - bilangan_2
perkalian = bilangan_1 * bilangan_2
pembagian = bilangan_1 / bilangan_2
pembagian_bulat = bilangan_1 // bilangan_2
modulus = bilangan_1 % bilangan_2

print("\n=== HASIL PERHITUNGAN ===")
print(f"{bilangan_1} + {bilangan_2} = {penjumlahan}")
print(f"{bilangan_1} - {bilangan_2} = {pengurangan}")
print(f"{bilangan_1} * {bilangan_2} = {perkalian}")
print(f"{bilangan_1} / {bilangan_2} = {pembagian}")
print(f"{bilangan_1} // {bilangan_2} = {pembagian_bulat}")
print(f"{bilangan_1} % {bilangan_2} = {modulus}")
```

```python
# Nama            : Kharis Ramadila
# NIM             : 4232501033
# Tanggal         : 27 September 2026
# Tujuan Program  : Menentukan apakah nilai ujian termasuk
#                   kategori lulus menggunakan operator relasional.

print("=== PROGRAM PENENTUAN KELULUSAN ===")

nilai_ujian = float(input("Masukkan nilai ujian (0-100): "))

lulus = nilai_ujian >= 60

print(f"Nilai yang dimasukkan adalah {nilai_ujian}.")
print(f"Nilai ini termasuk kategori lulus (>= 60): {lulus}")
```

```python
# Nama            : Kharis Ramadila
# NIM             : [Isi NIM]
# Tanggal         : 27 September 2026
# Tujuan Program  : Menentukan apakah seseorang boleh mengemudi
#                   berdasarkan umur dan kepemilikan SIM.

print("=== PROGRAM KELAYAKAN MENGEMUDI ===")

umur = int(input("Masukkan umur: "))
input_sim = input("Apakah memiliki SIM? (True/False): ")

# Mengubah masukan teks menjadi nilai Boolean.
punya_sim = input_sim.strip().lower() == "true"

# Seseorang boleh mengemudi apabila berumur minimal 17 tahun
# dan memiliki SIM.
boleh_mengemudi = (umur >= 17) and punya_sim

# Latihan operator precedence.
# Tanda kurung memastikan pengurangan dilakukan terlebih dahulu,
# kemudian hasilnya dikalikan 2.
skor_umur = (umur - 17) * 2

print("\n=== HASIL PEMERIKSAAN ===")
print(f"Umur: {umur} tahun")
print(f"Memiliki SIM: {punya_sim}")
print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")
print(f"Skor latihan operator precedence: {skor_umur}")
```

```python
# Nama            : Kharis Ramadila
# NIM             : 4232501033
# Tanggal         : 27 September 2026
# Tujuan Program  : Menghitung efisiensi termal unit pembangkit
#                   berdasarkan energi input dan energi output.

print("=== PROGRAM EFISIENSI TERMAL ===")

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print("\n=== HASIL PERHITUNGAN ===")
print(f"Energi input  : {energi_input:.2f} MJ")
print(f"Energi output : {energi_output:.2f} MJ")
print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%.")
```

```python
# Nama            : Kharis Ramadila
# NIM             : 4232501033
# Tanggal         : 27 September 2026
# Tujuan Program  : Menentukan apakah unit pembangkit memerlukan
#                   maintenance berdasarkan suhu boiler dan beban operasi.

print("=== PROGRAM PEMERIKSAAN MAINTENANCE ===")

suhu_boiler = float(input("Masukkan suhu boiler (derajat Celsius): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(
    f"Pada suhu boiler {suhu_boiler:.2f} derajat Celsius "
    f"dan beban operasi {beban_operasi:.2f}%, "
    f"status perlu maintenance adalah {perlu_maintenance}."
)
```

```python
A1
Operator precedence adalah aturan urutan pengerjaan operasi dalam Python. Pada ekspresi:
10 + 5 * 2
perkalian memiliki prioritas lebih tinggi daripada penjumlahan. Oleh karena itu, Python menghitung 5 * 2 = 10
terlebih dahulu, kemudian 10 + 10 = 20. Jika ingin penjumlahan dikerjakan lebih dahulu, digunakan tanda kurung:
(10 + 5) * 2
Sumber: Dokumentasi resmi Python
https://docs.python.org/3/reference/expressions.html#operator-precedence
```

```python
A2
Operator relasional digunakan untuk membandingkan 2 nilai dan menghasilkan True atau False. Contohnya adalah >, , >=, <=, ==, dan !=.
Operator logika digunakan untuk menggabungkan beberapa kondisi. Operator logika terdiri atas and, or, dan not.
Contohnya, seseorang boleh mengemudi apabila berumur minimal 17 tahun dan memiliki SIM:

umur = 18
punya_sim = True
boleh_mengemudi = (umur >= 17) and punya_sim
print(boleh_mengemudi)

Operator >= membandingkan umur, sedangkan and menggabungkan syarat umur dan kepemilikan SIM.
Hasilnya adalah True karena kedua syarat terpenuhi.
```

```python
A3
Salah satu kesalahan yang sering terjadi adalah menggabungkan operator or dan and tanpa tanda kurung. Contohnya:
boleh_masuk = nama == "Alex" or nama == "John" and umur >= 2

Python mengerjakan and lebih dahulu, sehingga syarat umur hanya berlaku untuk John. Akibatnya, Alex tetap diperbolehkan masuk
meskipun belum memenuhi syarat umur. Kode tersebut seharusnya diperbaiki dengan tanda kurung:
boleh_masuk = (nama == "Alex" or nama == "John") and umur >= 2

Dengan begitu, syarat umur berlaku untuk Alex dan John.

Sumber: https://www.geeksforgeeks.org/python/precedence-and-associativity-of-operators-in-python/
```
