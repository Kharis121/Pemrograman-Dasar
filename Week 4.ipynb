```python
# Konversi total detik ke jam, menit, dan detik

total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print("Hasil konversi: %d jam, %d menit, %d detik" % (jam, menit, detik))
```

```python
# Menghitung nilai akhir mahasiswa dari 5 komponen

nama = input("Masukkan nama mahasiswa: ")
tugas = float(input("Masukkan nilai tugas: "))
kuis = float(input("Masukkan nilai kuis: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))
praktikum = float(input("Masukkan nilai praktikum: "))
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

# Menghitung nilai akhir
nilai_mahasiswa = (
    (0.20 * tugas) +
    (0.10 * kuis) +
    (0.25 * uts) +
    (0.30 * uas) +
    (0.15 * praktikum)
)

# Menentukan kelulusan
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

# Menampilkan hasil
print("\nNama Mahasiswa:", nama)
print("Nilai Akhir: %.2f" % nilai_mahasiswa)
print("Standar Kelulusan: %.2f" % standar_kelulusan)
print("Apakah Mahasiswa Lulus?", apakah_lulus)
print("Perlu Remedial?", perlu_remedial)
```

```python
# Menghitung total harga setelah diskon

belanja = float(input("Masukkan total belanja: Rp"))

diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon

print("Total belanja : Rp%.2f" % belanja)
print("Total diskon  : Rp%.2f" % diskon)
print("Total bayar   : Rp%.2f" % total_harga)
```

```python
#Menentukan tahun kabisat

tahun = int(input("Masukkan tahun: "))

kabisat = ((tahun % 4 == 0) and (tahun % 100 != 0)) or (tahun % 400 == 0)

print("Tahun yang dimasukkan:", tahun)
print("Apakah termasuk tahun kabisat?", kabisat)
```

```python
#Menghitung total harga pembelian

nama_barang = input("Masukkan nama barang: ")
harga_barang = float(input("Masukkan harga barang: Rp"))
jumlah_barang = int(input("Masukkan jumlah barang: "))

total_harga = harga_barang * jumlah_barang

print("\nNama barang   :", nama_barang)
print("Harga barang  : Rp%.2f" % harga_barang)
print("Jumlah barang :", jumlah_barang)
print("Total harga   : Rp%.2f" % total_harga)
```
