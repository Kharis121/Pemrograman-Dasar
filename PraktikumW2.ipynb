```python
#Latihan 5

spesifikasi_motor = {
    'nama': 'Motor Induksi 3 Fasa',
    'daya_kW': 15.0,
    'tegangan_V': 380,
    'rpm': 1450}

# Cetak salah satu nilai lewat key-nya
print(spesifikasi_motor['tegangan_V'])

# Ubah nilai rpm (pakai =)
spesifikasi_motor['rpm'] = 1480

# Cetak ulang seluruh dictionary
print(spesifikasi_motor)
```

```python
# Latihan 6 — Gabungan (List of Dictionary)

# Membuat list yang berisi 3 dictionary unit pembangkit
pembangkit = [
    {"nama": "PLTU Jaya", "kapasitas_MW": 65, "status": "Operasi"},
    {"nama": "PLTA Sukses", "kapasitas_MW": 126, "status": "Operasi"},
    {"nama": "PLTP Diamond", "kapasitas_MW": 60, "status": "Pemeliharaan"}
]

# (1) Cetak nama unit kedua (indeks 1, lalu panggil key 'nama')
print(pembangkit[1]['nama'])

# (2) Ubah status salah satu unit (misal: unit ketiga dari 'Pemeliharaan' jadi 'Operasi')
pembangkit[2]['status'] = 'Operasi'

# (3) Tambahkan satu unit baru ke list dengan append()
pembangkit.append({'nama': 'PLTG Sinar', 'kapasitas_MW': 100, 'status': 'Standby'})

# (4) Cetak seluruh data
print(pembangkit)
```
