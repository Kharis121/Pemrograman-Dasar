```python
#Soal 1 : Belanja Buku

# Data
jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000

# Perhitungan
total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon
kembalian = uang_budi - bayar

# Patungan 4 orang
per_orang = bayar // 4
sisa = bayar % 4

# Cek uang cukup
cukup = uang_budi >= bayar

# Output
print("Total      :", total)
print("Diskon     :", diskon)
print("Bayar      :", bayar)
print("Kembalian  :", kembalian)
print("Per orang  :", per_orang)
print("Sisa       :", sisa)
print("Uang cukup?:", cukup)
```

```python
#Soal 2 : PLTS Atap Rumah

# Data
jumlah_panel = 12
daya_panel = 450      # Wp
jam_matahari = 5      # jam/hari
efisiensi = 0.8
kebutuhan = 25        # kWh/hari
tarif = 1444.70       # Rp/kWh

# Perhitungan energi harian
energi_harian = (jumlah_panel * daya_panel * jam_matahari * efisiensi) / 1000

# Energi bulanan (30 hari)
energi_bulanan = energi_harian * 30

# Nilai listrik yang dihasilkan
nilai_listrik = energi_bulanan * tarif

# Kekurangan energi
kekurangan = kebutuhan - energi_harian

# Cek apakah mencukupi
mencukupi = energi_harian >= kebutuhan

# Output
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
#Soal 3 : Generator Diesel dan Rugi Kabel

# Data
daya_generator = 200      # kW
jam_operasi = 8           # jam/hari
konsumsi_solar = 0.25     # liter per kWh
harga_solar = 13500       # Rp/liter
kapasitas_tangki = 1000   # liter

arus = 20                 # A
hambatan = 0.5            # ohm

# Perhitungan energi
energi = daya_generator * jam_operasi

# Solar per hari dan biaya
solar_harian = energi * konsumsi_solar
biaya = solar_harian * harga_solar

# Lama tangki cukup
hari_cukup = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

# Rugi daya kabel
rugi_kabel = (arus ** 2) * hambatan

# Cek rugi tinggi
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

# Output
print("Energi (kWh)   :", energi)
print("Solar (liter)  :", solar_harian)
print("Biaya (Rp)     :", biaya)
print("Tangki cukup   :", hari_cukup, "hari")
print("Sisa solar (L) :", sisa_solar)
print("Rugi kabel (W) :", rugi_kabel)
print("Rugi tinggi?   :", rugi_tinggi)
```
